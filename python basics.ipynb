#Operators:
1.Arithmetic operator:divide,multiply,add,modulus,exponents,floor divison


```python
#Arithmetic operators
a=10
b=20
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)#modulus
print(a**2)#exponent
print(10//3)#float division
```

```python
#2.Assignment operators:assign value to any variable 
x=5
print(x)
x=x+3
print(x)
x+=3
print(x)
a=4
a-=3
print(a)

```

```python
#3.Comparison operators:compare values
a=10
b=20
print(a==b)
print(a!=b)
print(a<b)
print(a>b)
print(a<=b)
print(a>=b)
```

```python
#4.Logical operators:combine conditional statements
#and:return true if both statements are true
#or:return true if one of the statements is true
#not:reverse the result,return false if the result is true

a=10
b=12
print(a!=5 and a<b )
print(a>b or a<b)  
print(not(a>b and a<15))
print(not(a>5 or a<15))
```

```python
#5.Membership operators:membership operators are used to test if a sequence is presented in an object
str1="HELLO"
print("H" in str1)
print("h" not in str1)#it is a case sensitive
l=[10,20,30,40]
print(50 not in l)
```

```python
#6.Identity operators:compare the objects,not if they are equal, but if they are actually the same object,with the same memory location
a=10
b=10
print(a is b,a==b)
print(a is not b,a!=b)
```

```python
#7.Bitwise operators:used to compare binary numbers
#and:sets each bit to 1 if both bits are 1()
#or:sets each bit to 1 if one of two bits is (|)
#xor:sets each bit to 1 if only one of two bits is 1(^)
#not:inverts all the bits(~)
y=10
z=8

print(bin(y),bin(z))
print(y&z,bin(y&z))
print(y|z,bin(y|z))
print(y^z,bin(y^z))
print(~y,bin(~y))
```
