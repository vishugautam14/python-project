```python
#LIST:it is mutable,ordered,allows duplicate elements,can contain different data types,can be nested,can be sliced and indexed,can be iterated over,can be modified (add/remove elements),supports various built-in methods (append, extend, insert, remove, pop, clear, sort, reverse),can be concatenated and repeated.Itcan store different data types, including other lists, making it versatile for various applications. Lists are defined using square brackets [] and can be created with or without initial elements. They are commonly used for storing collections of items, such as numbers, strings, or objects, and can be easily manipulated through indexing and slicing.
#List indexing and slicing 
l=['apple', 'banana', 20,30, 'elderberry']
print(l[0])
print(l[1:4])
print(l[-1][5])
print(l[-1::-1])#reverse the list 
l=['apple', 'banana', 20,30, 'elderberry',[10,20,30]]#add an inner list 
print(l[-1][1])

l=[10,20,30,40,50,60]
a=len(l)
#for i in range (a):
 #   print(i,l[i])

for n in l:
    print(n)
```

```python
#Functions of list 
l=[1,2,3,4]
l.append(5)#add an item at the end of the list
print(l)
l.insert(2,10)#item at at the postiton 2
print(l)
l.extend([6,7,8])#add the item at the end and if we do it in append so the whole lsit will add at the last of the list 
print(l)

n=[1,3,6,7,8,4,5]
print(n.pop())#remove the last item of the list and return it and tells which item is removed 
print(n)#gives the list after removing the last item
print(n.pop(1))#remove the item at position 1 and return it
print(n)
print(n.remove(4))#remove the item 4 from the list and return None
print(n)

a=[10,33,78,25,45,33,16,28]
a.sort()#sort the list in ascending order
print(a)
a.sort(reverse=True)#sort the list in descending order
print(a)

a.reverse()
print(a)

print(a.count(33))
a.clear()
print(a)
```
