#Data type-the type of data

there are two types of data type 

1.mutable-which state can be changed or contennt can be changed 

-list

-dictionary

-byte array 

2.Immutable-which content cannot be changed 

-integer

-float

-string

-complex

-tuple 

-set 



```python
#Number 
#1.integer
#2.float
#3.complex

a=2
print(a, 'is a type of:',type(a))

b=2.4
print(b,'is a type of:',type(b))

c=1+3j
print(c, 'is a type of:',type(c))
```

```python
#String:one or more characters in single, double and triple quotations
d='hello'
print(d,'is a type of:',type(d))
c='''hello
my name is Vishu'''
print(c,type(c))
```

```python
#List: A list is a collection which is ordered and changeable. Allows duplicate members,denoted by[]

a=[1,2.2,'hello']
a[2]='hi'
print(a,type(a))
```

```python
#Tuple: A tuple is a collection which is ordered and unchangeable. Allows duplicate members,denoted by().Its is faster than list and used to store data which is not going to change.
t=(1,'hi',1.1)
t=(1)
print(t,type(t))
```

```python
#Dictionary: A dictionary is a collection which is unordered, changeable and indexed. No duplicate members,denoted by{}.It is used to store data values in key:value pairs.
d={'name':'vishu','age':20}
print(d['name'])
print(d,type(d))
```

```python
#Set: A set is a collection which is unordered and unindexed. No duplicate members,denoted by{}.It is used to store multiple items in a single variable.its is immtuable and have unique value 
s={1,2,3,1}
print(s,type(s))
```
