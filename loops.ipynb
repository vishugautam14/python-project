1.for loops:A for loop is used to iterate the sequences(list,tuples,strings or range of no.)

```python
for a in range(5):
    print(a)

for n in range(1, 6):
    print(n,'welcome')

for i in range(1,11,2):
    print(i)    
```

```python
#decrement

for i in range(5, 0, -1):
    print(i)
#start=5
# stop>0=1 
# step=-1   

for a in range(22, 0, -2):
    print(a)
```

```python
#string in for loop
n='hello'
for s in n:
    print(s)

#list in for loop    
l=[1,2,3,4,5]
for i in l:
    print(i)
```

```python
#Nested loop:means placing for loop inside another for loop
#for outer in range(x):
#    for inner in range(y):
#        print(outer, inner)

for i in range(1,6):
    print(i,end=' ')
print(' ')    
for i in range(1,6):
    print(i,end=' ')
print(' ')    
for i in range(1,6):
    print(i,end=' ')
print(' ')    
#in this one task is repeating again and again that is print(i,end=' ') is repeating 3 times so we can use nested loop to avoid this repetition
```

```python
for r in range(1,6):
    for i in range(1,6):
       print(i,end=' ')
    print(' ') 
#the outer for loop is used to repeat the inner for loop 5 times and the inner for loop is used to print the numbers from 1 to 5 in a single line. The print(' ') statement is used to print a new line after each iteration of the outer for loop.    
```

```python
for i in range(6,11):
   
    for j in range(1,11):
        print(i,'*',j,'=',i*j)
    print(' ')
```

2.while loop:it repeats as long the condition is true.

-it only works on no.


```python
start=1
while start<=10:
    print(start)
    start+=1
```
