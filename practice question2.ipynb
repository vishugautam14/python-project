```python
#Adhar card 16digit validation in string

userInput=input("enter your adhar card number")
if userInput.isdigit() and len(userInput)==16:
     print("valid adhar no.")
else:
    print("invalid adhar no.")     
```

```python
#pan card no. in string

userInput=input("enter your pan card number")
panNum=userInput.upper()
if len(panNum)==10 and panNum[0:5].isalpha() and panNum[5:9].isdigit() and panNum[-1].isalpha():
    print("valid pan card number")
else:
    print("invalid pan card number")    
```

```python
#vowels count in string
userInput=input("enter your pan card number")
vowels="aeiouAEIOU"
count=0
for a in userInput:
    if a in vowels: 
        print(a)
        count+=1
print("vowels:",count)        
```
