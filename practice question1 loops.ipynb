```python
#sum of 10 natural numbers 
sum=0
for i in range(1,11):
    sum=sum+i
    
print('the sum of first 10 natural number is:',sum)

```

```python
#check whether the entered number is a perfect number or not
num=eval(input('enter the number:'))
pertotal=0
for i in range(1,num):
    if num%i==0:
        pertotal=pertotal+i
print(num,pertotal)
       
if pertotal==num:
    print(num,'is a perfect number')
else:
    print(num,'is not a perfect number')    

```

```python
#check whether the entered number is a prime number or not
num=eval(input('enter the number:'))
primestatus=True
midvalue=num//2

for n in range(2,midvalue):
    if num%n==0:
        primestatus=False
        break
if primestatus==True:
    print(num,'number is prime')
else:
    print(num,'number is not prime')         
    
```

```python
#reverse a number using while loop
num=eval(input('enter the number:'))
rev=0
print(num)
while num>0:
   lastdigit=num%10
   rev=rev*10+lastdigit
   num=num//10
print(rev)
```

```python
#count the digit in a number
num=eval(input('enter the number:'))
rev=0
c=0
print(num)
while num>0:
   c=c+1
   lastdigit=num%10
   rev=rev*10+lastdigit
   num=num//10
print(rev)
print(c)
```

```python
#right angle traingle of a star
n=5
for i in range(1,n+1):
    for j in range(1,i+1):
        print('*',end='')
    print()
```
