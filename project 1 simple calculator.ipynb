```python
print('simple calculator')
a=int(input('enter first number:'))
b=int(input('enter second number:'))

oper=input('enter operation(+,-,*,/):')
if oper=='+':
   print(a+b)
elif oper=='-':
   print(a-b)
elif oper=='*':
   print(a*b)
elif oper=='/':
   print(a/b)
else:
   print("Invalid operation")



```
