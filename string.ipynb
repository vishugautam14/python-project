```python
# slicing in string 
w="welcome"
print(w[2])
print(w[-5])#from the reverse start from -1
print(w[0:3])
print(w[::5])#5 character gaps in this 
print(w[:5])#start defaulting from 0  
print(w[0::])#end to last default 
print(w[0::4])#4 character gaps in this 
print(w[::2])

print(w[-1])#last character
print(w[-1::-1])#reverse the string
print(w[-1:2:-1])#reverse the string from last to 2nd index
print(w[-4:-1])#reverse the string from 4th to last index to last index
print(w[-4::])#reverse the string from 4th index to the end
```

```python
#to read all the characters in the string
w="welcome"
t=len(w)
print(t)
for i in range(t): #range(0,6)
    print(i,w[i])

#in reverse 
for a in range(t-1,-1,-1): #range(6,-1,-1)
    print(a,w[a])


#direct iterate 
# fro i in w:
# print(i)    


```

```python
#Functions of string :string is immutable that is it cannot be changed once created

s=" welcome TO ws"
n=s.lower()#all the characters in the string will be converted to lower case
print(n)
print(s.upper().strip())#all the characters in the string will be converted to upper case
print(s.title())##all the first character of the string will be converted to upper case
print(s.strip())#it will remove the white spaces from the string
print(s.replace("TO","HI"))#it will replace the string with the new string
print(s.split(" "))#it will split the string into list of words
print(s.split("e"))

w="welcome"
print(len(w))#it will return the length of the string
print(w.find("e"))#it will return the index of the first occurrence of the string
print(w.count("e"))#it will return the number of occurrences of the string
print(w.find("z"))
print(w.index("e"))#it will return the index of the first occurrence of the string
#print(w.index("z")):it will return the index of the first occurrence of the string but if the string is not found it will raise an error
```
